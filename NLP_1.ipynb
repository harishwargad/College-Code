```python
import nltk
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from collections import Counter
import string
```

```python
text = """The crowd yells and screams for more memes.
He waited for the stop sign to turn to a go sign.
Instead of a bachelorette party
My dentist tells me that chewing bricks is very bad for your teeth.
Car safety systems have come a long way, but he was out to prove they could be outsmarted."""

print("original text: ", text)
```

```python
words = word_tokenize(text)
print("word tokenized: ", words)

sentences = sent_tokenize(text)
print("sentence tokenized: ", sentences)
```

```python
lower_words = []
for word in words:
    lower_words.append(word.lower())

print('lowercase words: ', lower_words)
```

```python
clean_words = []
for word in lower_words:
    if word not in string.punctuation:
        clean_words.append(word)

print('cleaned words: ', clean_words)
```

```python
stop_words = set(stopwords.words('english'))

filtered_words = []
for word in clean_words:
    if word not in stop_words:
        filtered_words.append(word)

print('filtered words: ', filtered_words)
```

```python
stemmer = PorterStemmer()

stemmed_words = []
for word in filtered_words:
    stemmed_words.append(stemmer.stem(word))

print('stemmed words: ', stemmed_words)
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_words = []
for word in filtered_words:
    lemmatized_words.append(lemmatizer.lemmatize(word))

print('lemmatized words: ', lemmatized_words)
```

```python
frequency = Counter(filtered_words)

for word, count in frequency.items():
    print(f"{word} : {count}")
```
